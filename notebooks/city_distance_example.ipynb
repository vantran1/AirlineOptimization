```julia
using CSV, DataFrames, Plots, Geodesics, GeoMakie
using LaTeXStrings, StatsPlots, LinearAlgebra
using GLMakie
```

```julia
city_coords = CSV.read("data/realcities.csv", DataFrame);
C = Matrix(city_coords);
```

```julia
function distance_matrix(C::Matrix)
    n = size(C,1)
    D = zeros((n,n))
    
    a, f = Geodesics.EARTH_R_MAJOR_WGS84, Geodesics.F_WGS84
    # Unit conversion factor: 1 nmi = 1852 meters
    conv_fac = 1/1852
    for i in 1:n
        for j in 1:n
            loni, lati = C[i,2], C[i,1]
            lonj, latj = C[j,2], C[j,1]
            dist, _ , _ = Geodesics.inverse(deg2rad.((loni, lati, lonj, latj))..., a, f)
            dist_nmi = dist * conv_fac
            D[i,j] = dist_nmi
        end
    end
    return D
end
```

```julia
D = distance_matrix(C);
```
